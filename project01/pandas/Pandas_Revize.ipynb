## Veriyi Filtreleme

Zaman zaman verinin tümüyle çalışmak yerine, veri içerisinden belirli bir kısımla çalışma ihtiyacı duyarız. Bu sebeple, verilerimize çeşitli filtreleme ve maskeleme işlemleri uygularız.

```python
mask = (df.calories > 100)
```

```python
mask
```

Ancak filtreleme işlemi sonucunda elde ettiğimiz değerleri bir `DataFrame` içerisinde gösterirsek yalnızca `True` dönen satırları DataFrame'de tutmuş oluruz.

```python
df[mask].head(10)
```

Tabi ki bir değişken tanımlaması yapmadan da bu tarz işlemleri yapmak mümkün :)

```python
df[df.calories > 100].head(10)
```

Verimiz üzerinde, birden fazla filtre uygulamak istersek de, `and (&)`, `or (|)` operatörlerini kullanmak istediğimiz sonuca kolayca ulaşmamıza sağlayacaktır.

```python
mask = (df.calories > 100) & (df.sugars < 8) & ((df.mfr == 'K') | (df.mfr == 'G'))
```

```python
df[mask]
```

## Veriye Fonksiyonlar Uygulama

```python
df.vitamins.value_counts()
```

`value_counts()` fonksiyonundan elde ettiğimiz sonuca göre `vitamins` sütunumuz sadece 3 farklı sayısal değer barındırmakta. Bu senepten dolayı, hadi `vitamin_at_least_25` adında yeni bir sütun oluşturalım.

```python
df['vitamin_at_least_25'] = df.vitamins >= 25
```

```python
df.head()
```

### Apply()

Peki, veri setimiz içerisindeki ürünlerin vitamin değerlerine göre onları gruplamak istesem? Bunu, `.apply()` fonksiyonunu kullanarak rahatlıkla yapabilirim.

```python
def get_vitamin_level(vitamins):
    if vitamins > 99:
        return 'high'
    elif vitamins > 24:
        return 'medium'
    else:
        return 'low'
```

```python
get_vitamin_level(100)
```

```python
get_vitamin_level(25)
```

```python
df.vitamins.apply(get_vitamin_level)
```

```python
df['vitamin_level'] = df.vitamins.apply(get_vitamin_level)
```

```python
df.head()
```

Buna benzer işlemleri `def` le fonksiyon tanımlamadan `lambda` fonksiyonunda yararlanarak da yapabilirdik :)

```python
df['vitamin_perc'] = df.vitamins.apply(lambda x: str(x)+'%')
```

```python
df.head()
```

### Map()

`.apply()` bizim için çok kullanışlı ve sevdiğimiz bir fonksiyon, aynı `.map()` fonksiyonu gibi :)

```python
mfr_dict = {
    'A': 'Am Home Food Prods',
    'G': 'General Mills',
    'K': 'Kelloggs',
    'N': 'Nabisco',
    'P': 'Post',
    'Q': 'Quaker Oats',
    'R': 'Ralston Purina'
}
```

Tek yapmamız gereken tanımladığımız `dictionary` yapısını elimizdeki `DataFrame` ile birleştirmek :)

```python
df.mfr.map(mfr_dict)
```

```python
df['manufacturer'] = df.mfr.map(mfr_dict)
```
